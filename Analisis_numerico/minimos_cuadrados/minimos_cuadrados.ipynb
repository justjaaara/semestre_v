### Qué es cada variable:

- n: cantidad de datos
- Sx: Sumatoria de todas las x
- Sy: Sumatoria de todas las y
- Sxy: La sumatoria de la multiplicación de x*y (x0y0 +x1y1...xnyn)
- Sx²: Sumatoria de cada x²
- [Sx]²: Sumatoria de todas las x y elevar el resultado al cuadrado

```python
import sys
import numpy as np
import matplotlib.pyplot as plt

sys.path.append('/Analisis_numerico/minimos_cuadrados')
from minimos_cuadrados_funcion import minimos_cuadrados

x  = np.array([0,2,3,6,7])
y = np.array([0.120, 0.153, 0.170, 0.225, 0.260])

a0, a1 = minimos_cuadrados(x, y)

plt.plot(x, y, 'o', label='Datos')
plt.plot(x, a0 + a1*x, label='Ajuste lineal')
plt.xlabel('x')
plt.ylabel('y')
plt.title('Ajuste lineal')
plt.legend()
plt.grid()
print(f"Polinomio: {a0} + {a1}x")
```
