```python
# cambiando solamnete la ecuación diferencial
def f(t, u):
    """
    Parametros:
    t: tiempo
    u: arreglo de variables u=[w,z] es decir que f(t, [w,z])
    """
    w= u[0]
    z= u[1]
    n = len(u)
    y= np.zeros(n)
    y[0] = z
    y[1] = t*np.exp(t)-t+2*z-w
    return y

from metodo_euler import metodo_euler_profe

tiempo, resultado = metodo_euler_profe(f,0,1,0.25, [0,0])

for w_n, z_n in resultado:
    print(f"{w_n} |  {z_n}")
```
