Solucionar el sistema de ecuaciones diferenciales de primer orden por el método de Euler para $c=0.5, \quad c=1.5, \quad c=2.2, \quad c=3$.

```python
import numpy as np
# SOLUCIONANDO PARA C= 0.5
def func(c):
    def emociones(t,y):
        x0, y0 = y[0], y[1]
        
        xprima = y0 + np.sin(c*t)
        yprima = -4*x0

        return np.array([xprima, yprima])

    return emociones
```

```python
from metodo_euler import metodo_euler_profe
from matplotlib import pyplot as plt    
import numpy as np
cs = [0.5, 1.5, 2.2, 3]
for c in cs:
    tiempo, modelo = metodo_euler_profe(func(c),0, 30, 0.01, [0,0])
    denise = np.array(modelo)[:,0]
    craig = np.array(modelo)[:,1]
    plt.figure(figsize=(12, 4))
    plt.subplot(131)
    plt.plot(tiempo, denise, label='x', color='blue')
    plt.subplot(132)
    plt.plot(tiempo, craig, label='y', color='green')
    plt.subplot(133)
    plt.plot(denise, craig, label='x-y', color='red')
    plt.show()
```
