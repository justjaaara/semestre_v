```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt
x = sp.symbols('x')

def lagrange(x_data, y_data):
    
    P = 0 # Acumulador del polinomio
    n = len(x_data) # Número de datos
    
    for i in range(n):
        Li = 1
        for j in range(n):
            if j != i:
                Li *= (x - x_data[j]) / (x_data[i] - x_data[j])
        P += Li * y_data[i]
    return sp.expand(P)

x_data = np.array([0, 0.6, 0.9])
y_data = np.array([1, 1.264911, 1.378404])

P = lagrange(x_data, y_data)
P = sp.lambdify(x, P)

ux = np.linspace(min(x_data), max(x_data), 100)
plt.plot(x_data, y_data, 'gp', label='Datos observados')
plt.plot(ux, P(ux), 'r', label='Polinomio de Lagrange')
plt.legend()
        
# Comparando con polinomio simple

```
