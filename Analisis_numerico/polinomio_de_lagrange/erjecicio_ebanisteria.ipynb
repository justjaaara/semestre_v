```python
import numpy as np
import sympy as sp
import matplotlib.pyplot as plt
x = sp.symbols('x')

def lagrange(x_data, y_data):
    
    P = 0 # Acumulador del polinomio
    n = len(x_data) # Número de datos
    
    for i in range(n):
        Li = 1
        for j in range(n):
            if j != i:
                Li *= (x - x_data[j]) / (x_data[i] - x_data[j])
        P += Li * y_data[i]
    return sp.expand(P)

x_data = np.array([17, 19, 20 ,22, 23, 25, 31, 32, 33, 36 ,37 ,38, 39, 41])
y_data = np.array([19, 25, 32, 51, 57, 71, 141, 123, 187, 192, 205, 252, 248, 294])
P = lagrange(x_data, y_data)
print("Polinomio de Lagrange:")
print(P)
P = sp.lambdify(x,P)

#Aproximando P(17) y P(28)
print("P(17) = ", P(17))
print("P(28) = ", P(28))


#Grafica de disperssión de puntos
ux = np.linspace(min(x_data), max(x_data), 100)
plt.plot(x_data, y_data, 'ro', label='Datos observados')
plt.plot(ux, P(ux), label='Polinomio de Lagrange')
plt.legend()
plt.grid()
plt.legend()
plt.show()
```
