```python
import numpy as np
import matplotlib.pyplot as plt

x_d = np.array([36, 67.25, 93, 483.80, 887.97, 1764.50, 2791.05, 3656.90]) # Distancia en millones y es x_d porque es la que tengo
y_d = np.array([88, 224.7, 365.3, 4331.8, 10760.0, 30684.0, 60188.3, 90466.8]) # Periodo y es la y_d porque es la que voy a predecir

plt.plot(x_d, y_d, 'o', label='Datos')
plt.grid()
plt.xlabel('Distancia (Miles de millones)')
plt.ylabel('Periodo')
plt.legend()
```

Ahora verifico qué modelo se puede ajustar mejor

```python
from graficar_escalas import graficar_escalas

graficar_escalas(x_d, y_d)
```

Según lo anterior al parecer el modelo que más aplica es log(x), log(y)

```python
from minimos_cuadrados_funcion import minimos_cuadrados

a0, a1 = minimos_cuadrados(np.log(x_d), np.log(y_d))

print(f"a0: {a0} a1: {a1}")

ux = np.linspace(min(np.log(x_d)), max(np.log(x_d)))
P = lambda x: a0 + a1 * x
plt.scatter(np.log(x_d), np.log(y_d), label="Grafica del modelo")
plt.plot(ux, P(ux), 'r', label='Modelo lineal')
plt.legend()
```

Se ajusta prefecto por lo que será muy útil para cnstruir el modelo no lineal


### Graficando el modelo no lineal

### Partir de $Y=a_0+a_1x$, dependiendo de la escala en este caso $log(x), log(y)$ es el modelo no lineal es:

$$log(y) = a_0+a_1log(x)$$
$$y = e^{a_0 + log(x^{a_1})}$$
$$M = e^{a_0}*x^{a_1}$$

Con el polinomio M ya se pueden hacer predicciones para el caso específico

```python
u_x = np.linspace(min(x_d), max(x_d))
M = lambda x: np.exp(a0) * x**a1

plt.title("Modelo no lineal")
plt.plot(x_d,y_d, 'or', label="Datos")
plt.plot(u_x, M(u_x), 'b', label="Ajuste")
plt.legend()
```

Ya con el modelo no linea podemos hacer predicciones y solucionar el item solicitado para Marte

```python
periodo_marte = M(141.75)
print(f"El periodo predecido para el planeta Marte es: {periodo_marte}")
```
