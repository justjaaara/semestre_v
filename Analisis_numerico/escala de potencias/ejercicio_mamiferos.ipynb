```python
import numpy as np
import matplotlib.pyplot as plt

x_d = np.array([4, 25, 200, 300, 2000, 5000, 30000, 50000, 70000, 450000, 500000, 3000000])
y_d = np.array([660, 670,420, 300, 205, 120, 85, 70, 72, 38, 40, 48])


plt.plot(x_d, y_d, 'o', label='Datos')
plt.grid()
plt.xlabel('Peso')
plt.ylabel('Pulso cardiaco')
plt.legend()
```

Ahora se verifica qué modelo se ajusta más a nuestros datos

```python
from graficar_escalas import graficar_escalas

graficar_escalas(x_d, y_d)
```

```python
from minimos_cuadrados_funcion import minimos_cuadrados

a0, a1 = minimos_cuadrados(np.log(x_d), np.log(y_d))

print(f"a0: {a0} a1: {a1}")

ux = np.linspace(min(np.log(x_d)), max(np.log(x_d)))
P = lambda x: a0 + a1 * x
plt.scatter(np.log(x_d), np.log(y_d), label="Grafica del modelo")
plt.plot(ux, P(ux), 'r', label='Modelo lineal')
plt.legend()
```

En la gráfica anterior se observa que el modelo si se ajusta, ahora vamos a graficar el modelo con los datos observados

### Graficando el modelo no lineal

### Partir de $Y=a_0+a_1x$, dependiendo de la escala en este caso $log(x), log(y)$ es el modelo no lineal es:

$$log(y) = a_0+a_1log(x)$$
$$y = e^{a_0 + log(x^{a_1})}$$
$$M = e^{a_0}*x^{a_1}$$

Con el polinomio M ya se pueden hacer predicciones para el caso específico

```python
u_x = np.linspace(min(x_d), max(x_d))
M = lambda x: np.exp(a0) * x**a1

plt.plot(x_d,y_d, 'or', label="Datos")
plt.plot(u_x, M(u_x), 'b', label="Ajuste")
plt.legend()
```

De lo anterior entonces ya podemos hacer predicciones para el mamifero y resolver el item a continnuación:

```python
latidos_musarana_etrusca = M(2)
print(f"La musaraña etrusca tiene {latidos_musarana_etrusca} latidos")

latidos_musarana_etrusca_real = 1200

error = abs((latidos_musarana_etrusca - latidos_musarana_etrusca_real)/latidos_musarana_etrusca_real)   
print(f"El erorr del modelo para esta predicción es de: {error}")
```
