```python
import numpy as np
import matplotlib.pyplot as plt
import sympy as sp

def lagrange_pol(x_data, y_data):
    x = sp.symbols('x')
    
    P = 0 # Acumulador del polinomio
    n = len(x_data) # Número de datos
    
    for i in range(n):
        Li = 1
        for j in range(n):
            if j != i:
                Li *= (x - x_data[j]) / (x_data[i] - x_data[j])
        P += Li * y_data[i]
    return sp.expand(P)

pez_azul = np.array([14999, 150000, 250000, 275000, 270000, 280000, 290000, 650000, 1200000, 1500000, 2750000])
cangrejo = np.array([99999, 850000, 1330000, 2500000, 3000000, 3700000, 4400000, 4660000, 4800000, 4420000, 5000000])
anios = np.array([1940, 1945, 1950, 1955, 1960, 1965, 1970, 1975, 1980, 1985, 1990])

polinomio_lagrange = lagrange_pol(anios, pez_azul)

plt.plot(anios, pez_azul, 'o', label='Datos')
plt.grid()
plt.xlabel('Años')
plt.ylabel('Pez azul')
plt.legend()
```

# Modelo del pez azul

1. Escogeimos Log x vs y
2. Hacemos minimos cuadrados (x, log(y))
3. Obtenemeos a0,a1
4. Graficamos los datos en esacala recta
5. Modelo no lineal

```python
# Graficamos los datos en la escala escogida y el modelo lineal determinado por una regresión de mínimos cuadrados

import numpy as np
from minimos_cuadrados_funcion import minimos_cuadrados

a0, a1 = minimos_cuadrados(x_d,np.log(y_d))
print(f"a0: {a0} a1: {a1}")

ux = np.linspace(min(x_d), max(x_d))
P = lambda x: a0 + a1 * x
plt.scatter(x_d, np.log(y_d), label="Gráfica x vs log(y)")
plt.plot(ux, P(ux), 'r', label='Modelo lineal')
plt.legend()
```

# Graficar el modelo no lineal

### Partir de $Y=a_0+a_1x$, dependiendo de la escala en este caso $x , log(y), y$ es el modelo no lineal es:

$$ln(y) = a_0 + a_1*x$$
$$y= e^{a_0 + a_1*x}$$
$$M= e^{a_0 + a_1*x}$$

```python
u_X = np.linspace(min(x_d), max(x_d))
M = lambda x: np.exp(a0 + a1 * x)

plt.plot(x_d,y_d, 'or', label="Datos")
plt.plot(u_X, M(u_X), 'b', label="Ajuste lineal")
plt.legend()
```

# Graficar el modelo no lineal

### Partir de $Y=a_0+a_1x$, dependiendo de la escala en este caso $\sqrt{x} , y, y$ es el modelo no lineal es:

$$y = a_0 + a_1\sqrt{x}$$
$$M = a_0 + a_1\sqrt{x}$$

```python
u_X = np.linspace(min(x_d), max(x_d))
M = lambda x: a0 + a1 * np.sqrt(x)
plt.plot(x_d, y_d, 'or', label="Datos")
plt.plot(u_X, M(u_X), 'b', label="Ajuste lineal")
plt.legend()
```

Puede que no sea tan preciso este método así que podríamos intentar con otros para hacer predicciones más acertadas
